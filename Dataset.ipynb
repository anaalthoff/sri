```python
# Import da biblioteca
from datasets import load_dataset

# Carrega o conjunto de dados
dataset = load_dataset("divergente/noticias-govbr-ptbr-1")

print(dataset)
```

## 2) Percorrer o conjunto de dados

```python
cont = 0
for i, x in enumerate(dataset["train"]):
      if cont < 10:
        print("\n", x)
        cont = cont + 1
```

## 3) Gerar um arquivo texto (csv) com os dados do dataset usando ";" como separador

```python
import pandas as pd

# Lista para receber os dados selecionados
dados = []

# Percorre o dataframe
for i, x in enumerate(dataset["train"]):
      # Seleciona somente 50 documentos
      if i <= 50:
        # Adiciona o id e o documento
        dados.append({
            "id": i + 1,
            "documento": x["title"]
        })

# Cria o DataFrame para poder salvar os dados
df = pd.DataFrame(dados)

# Define o caminho completo para salvar o arquivo no Google Drive
gdrive_path = '/content/drive/MyDrive/Colab Notebooks/SRI/data/documentos.csv'

# Salva como CSV no Google Drive
df.to_csv(gdrive_path, index=False, encoding="utf-8", sep=';')
```
